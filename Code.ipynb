##Identify and Handle Missing Values

Purpose:

We perform this step to identify and distinguish different types of missing values in the dataset. Some values are genuinely missing while others represent the absence of a property feature.

```python
df = df_raw.copy(deep=True)

blank_missing = df.isnull().sum().sum()

literal_na = (
    df.select_dtypes(include=["object", "string", "category"])
    .eq("NA")
    .sum()
    .sum()
)

print("Genuinely missing cells:", blank_missing)
print("Literal NA values:", literal_na)
```

##Handle missing values

Purpose:



We perform this step because some missing-value indicators represent valid property characteristics rather than data-entry errors.

```python
garage_present = df["Garage Type"] != "NA"

for col in ["Garage Qual", "Garage Cond"]:
    df.loc[
        garage_present & (df[col] == "NA"), col
    ] = np.nan


absence_columns = [
    "Alley",
    "Bsmt Qual",
    "Bsmt Cond",
    "Bsmt Exposure",
    "BsmtFin Type 1",
    "BsmtFin Type 2",
    "Fireplace Qu",
    "Garage Type",
    "Garage Finish",
    "Garage Qual",
    "Garage Cond",
    "Pool QC",
    "Fence",
    "Misc Feature"
]

for col in absence_columns:
    df[col] = df[col].replace("NA", "None")

print("Feature absence values handled successfully!")
```

##Handle missing numerical values

Purpose:



We perform this step because missing Lot Frontage values can affect numerical analysis and machine-learning algorithms that require complete data.


Properties in different neighborhoods may have different lot-frontage characteristics. Therefore, replacing all missing values with a single overall median may ignore important neighborhood differences.

```python
neighborhood_median = df.groupby(
    "Neighborhood"
)["Lot Frontage"].transform("median")

df["Lot Frontage"] = df["Lot Frontage"].fillna(
    neighborhood_median
)

df["Lot Frontage"] = df["Lot Frontage"].fillna(
    df["Lot Frontage"].median()
)

print(
    "Missing Lot Frontage values:",
    df["Lot Frontage"].isnull().sum()
)
```

Purpose:

We perform this step to handle missing garage construction years while distinguishing houses without garages from houses with unknown garage construction dates.

For properties without garages, the garage construction year is not applicable. We use 0 as a special absence indicator.

For properties that have a garage but have an unknown construction year, we use the median garage construction year as an estimate.

This approach preserves all records and avoids treating the absence of a garage as an ordinary missing measurement.

```python
no_garage = df["Garage Type"] == "None"

df.loc[no_garage, "Garage Yr Blt"] = 0

garage_year_median = df.loc[
    ~no_garage, "Garage Yr Blt"
].median()

df.loc[~no_garage, "Garage Yr Blt"] = (
    df.loc[~no_garage, "Garage Yr Blt"]
    .fillna(garage_year_median)
)

print(
    "Missing Garage Year Built values:",
    df["Garage Yr Blt"].isnull().sum()
)
```

Purpose:

We perform this step to ensure that basement measurements are complete and logically consistent.

Basement area consists of finished and unfinished sections. The total basement area should equal the sum of its component areas.

If these measurements are imputed independently, the resulting values could contradict one another.

Therefore, we estimate missing component measurements first and calculate the corresponding total from their sum.

For properties confirmed to have no basement, missing basement bathroom counts can be replaced with zero.

```python
no_basement = df["Total Bsmt SF"] == 0

for col in ["Bsmt Full Bath", "Bsmt Half Bath"]:
    df.loc[no_basement, col] = (
        df.loc[no_basement, col].fillna(0)
    )


basement_area_cols = [
    "BsmtFin SF 1",
    "BsmtFin SF 2",
    "Bsmt Unf SF",
    "Total Bsmt SF"
]

all_basement_unknown = (
    df[basement_area_cols].isnull().all(axis=1)
)

for col in basement_area_cols[:3]:
    df.loc[all_basement_unknown, col] = (
        df.loc[all_basement_unknown, col]
        .fillna(df[col].median())
    )

df.loc[all_basement_unknown, "Total Bsmt SF"] = (
    df.loc[
        all_basement_unknown,
        basement_area_cols[:3]
    ].sum(axis=1)
)

print("Basement measurements handled successfully!")
```

Purpose:


We perform this step to eliminate remaining missing values so that the dataset can be used for analysis and subsequent preprocessing.

For numerical columns, median imputation provides an estimate that is less sensitive to extreme observations.

For categorical columns, we replace genuinely missing information with Unknown rather than assigning an unverified category.

This helps preserve the dataset's records and prevents missing values from causing errors during later processing.

```python
numerical_columns = df.select_dtypes(
    include=np.number
).columns

for col in numerical_columns:

    if df[col].isnull().sum() > 0:

        df[col] = df[col].fillna(
            df[col].median()
        )

categorical_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns

for col in categorical_columns:

    if df[col].isnull().sum() > 0:

        df[col] = df[col].fillna("Unknown")


print("Remaining missing values handled successfully!")
```

Purpose:


We perform this step to evaluate whether our missing-value treatment was successful.

We compare the number of missing values before and after cleaning to verify that the intended missing values have been handled.

We also check whether the dataset's dimensions remain unchanged and whether related features, such as basement measurements, remain logically consistent.

```python
missing_comparison = pd.DataFrame({
    "Before Cleaning": df_raw.isnull().sum(),
    "After Cleaning": df.isnull().sum()
})

display(
    missing_comparison[
        missing_comparison["Before Cleaning"] > 0
    ]
)

print(
    "Missing cells before cleaning:",
    df_raw.isnull().sum().sum()
)

print(
    "Missing cells after cleaning:",
    df.isnull().sum().sum()
)

print("Dataset shape:", df.shape)

basement_total = (
    df["BsmtFin SF 1"]
    + df["BsmtFin SF 2"]
    + df["Bsmt Unf SF"]
)

assert np.allclose(
    df["Total Bsmt SF"],
    basement_total
), "Basement area measurements are inconsistent!"

assert df.isnull().sum().sum() == 0, (
    "Some missing values remain!"
)

print("Missing value handling verified successfully!")
```

##Outlier Detection and Handling

Purpose:


The IQR method helps identify unusually high or low values in Lot Area, Gr Liv Area, and SalePrice that could affect statistical analysis.


```python
outlier_columns = ["Lot Area", "Gr Liv Area", "SalePrice"]

outlier_results = []
outlier_flags = pd.DataFrame(index=df.index)

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    )

    outlier_flags[col] = outliers

    outlier_results.append({
        "Column": col,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outliers.sum()
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```

Purpose:


Some expensive or large houses may be genuine, so we examine extreme values before removing or changing them to avoid losing useful information.


```python
outlier_rows = df.loc[
    outlier_flags.any(axis=1),
    ["Order", "Lot Area", "Gr Liv Area", "SalePrice"]
]

display(
    outlier_rows.sort_values(
        by="Lot Area",
        ascending=False
    ).head(10)
)
```

Purpose:


Logarithmic transformation reduces the influence of extreme values and makes skewed distributions more balanced while preserving the original property data.


```python
df["Lot_Area_Log"] = np.log1p(df["Lot Area"])

df["Gr_Liv_Area_Log"] = np.log1p(df["Gr Liv Area"])

df["SalePrice_Log"] = np.log1p(df["SalePrice"])

skewness_comparison = pd.DataFrame({
    "Feature": [
        "Lot Area",
        "Gr Liv Area",
        "SalePrice"
    ],

    "Before": [
        df["Lot Area"].skew(),
        df["Gr Liv Area"].skew(),
        df["SalePrice"].skew()
    ],

    "After": [
        df["Lot_Area_Log"].skew(),
        df["Gr_Liv_Area_Log"].skew(),
        df["SalePrice_Log"].skew()
    ]
})

display(skewness_comparison.round(3))

print("Dataset shape:", df.shape)
```

Select Numerical Features
